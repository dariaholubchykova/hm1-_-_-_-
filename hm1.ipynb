1.Ви —  продакт менеджер в компанії, яка займається мобільними додатками. Ваша модель монетизації — підписка. У травні 2019 року до вас приєдналося 100 підписників, у червні — 150, у липні — 170. Скільки ці користувачі принесли вам до початку травня 2021 року (за два роки), якщо щомісячний рівень відтоку становить 5%, а підписка коштує $10?

s1 = 100 приєднались в травні
s2 = 150 приєднались в червні
s3 = 170 приєднались в липні
q = 1 - 0.05 = 0.95
p = 10$
Дохід для травня (24 місяці)
R1 = p*s1*(1-q^n)/(1-q) = 10 * 100 *(1-0.95^24) / (1-0.95) = 14160.22
Дохід для травня (23 місяці)
R2 = p*s2*(1-q^n)/(1-q) = 10 * 150 * (1-0.95^23) / (1-0.95) = 20779.29
Дохід для травня (22 місяці)
R3 = p*s3*(1-q^n)/(1-q) = 10 * 170 * (1-0.95^22) / (1-0.95) = 22999.86
R(all) = R1 + R2 + R3 = 57939.37$

2. Ви — менеджер по роботі з продуктом в компанії, яка розробляє мобільні ігри. У січні 2020 року у вас було 2000 активних користувачів. Щомісяця кількість нових користувачів зростає на 3%, але водночас 2% існуючих користувачів перестають користуватися додатком. У результаті, ефективне зростання активних користувачів складає 1% щомісяця.
    
    Ваша мета — визначити кумулятивну кількість активних користувачів за весь період від січня 2020 року до кінця грудня 2021 року (тобто за 24 місяці), враховуючи щомісячний приріст 1%.
    
    **Питання**: Якою буде загальна кількість активних користувачів, накопичена за 24 місяці?

s = 2000 кількість активних користувачів на січень 2020
q = 1 + 0.01 = 1.01 кожен місяць кількість користувачів збільшується на 1%
R = s * (1-q^n) / (1-q) = 2000 * (1-1.01^24)/(1-1.01) = 53947 -загальна кількість активних користувачів, накопичена за 24 місяці 


Ви — менеджер з розвитку бізнесу в компанії, що займається створенням платформи для онлайн-курсів. У січні 2021 року ви залучили 500 нових користувачів. У лютому кількість нових користувачів зросла на 50%, у березні — ще на 20%. З квітня приріст нових користувачів стабілізувався на рівні 10% щомісяця. Скільки активних користувачів ви мали загалом за перше півріччя 2021 року?

s = 500 кількість активних користувачів на січень 2021
q1 = 1 + 0,5 = 1,5  - приріст за лютий
q2 = 1 + 0.2 = 1,2  - приріст за березень
q3 = 1 + 0.1 = 1,1  - приріст з квітня 
R1 = 500 * 1,5 = 750 - кількість активних користувачів на лютий 2021
R2 = 750 * 1.2 = 900 - кількість активних користувачів на березень 2021
R3 = 900 * (1-1.1^3)/(1-1.1^3) = 2979- кількість активних користувачів на квітень-червень 2021
R(all) = 500 + 750 + 900 + 2979 = 5129
